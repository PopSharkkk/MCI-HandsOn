### **Deteksi Outlier**

---

Memeriksa empat fitur utama, yaitu `Recency`, `Frequency`, `Monetary`, dan `Avg_Order_Value`, untuk melihat apakah terdapat outlier yang perlu ditangani sebelum data ini digunakan lebih lanjut.

```python
# Mendeteksi Outlier
# Pilih beberapa fitur numerikal penting untuk dicek
features_to_check = ['Recency', 'Frequency', 'Monetary', 'Avg_Order_Value']

plt.figure(figsize=(12, 8))
# Buat visualisasi box plot
sns.boxplot(data=df[features_to_check], palette='Set3')

plt.title('Outlier Detection using Box Plots', fontsize=15)
plt.ylabel('Value Range')
plt.show()
```

# **Visualisasi**

---

Bagian ini digunakan untuk menyampaikan pola yang sudah ditemukan kepada audience. Karena tujuannya untuk komunikasi, visualisasi yang ditampilkan hanya berfokus pada beberapa kolom yang relevan, bukan seluruh kolom yang ada di dataset.

### **Distribusi Recency**

---

Menampilkan sebaran nilai `Recency` (berapa lama sejak user terakhir belanja) untuk memberi gambaran seberapa aktif user secara umum, dihitung dari jarak hari sejak transaksi terakhir.

```python
sns.histplot(df["Recency"], kde=True)
plt.title('Distribusi Recency')
plt.xlabel('Recency (hari sejak transaksi terakhir)')
plt.show()
```

### **Distribusi Monetary**

---

Menampilkan sebaran nilai `Monetary` untuk memberi gambaran seberapa besar total pengeluaran user.

```python
sns.histplot(df["Monetary"], kde=True)
plt.title('Distribusi Monetary')
plt.xlabel('Monetary (total pengeluaran user)')
plt.show()
```

### **Hubungan Frequency dan Monetary**

---

Menampilkan hubungan antara frekuensi transaksi dan total pengeluaran user, untuk melihat apakah user yang lebih sering bertransaksi juga cenderung mengeluarkan lebih banyak.

```python
sns.scatterplot(data=df, x="Frequency", y="Monetary")
plt.title('Frequency vs Monetary')
plt.show()
```

### **Hubungan Frekuensi Transaksi dengan Rata-rata Pengeluaran**

---

Line chart ini menunjukan bagaimana rata-rata nominal belanja user berubah seiring makin seringnya mereka bertransaksi, untuk melihat pola perilaku belanja secara lebih jelas dibanding scatter plot sebelumnya

```python
avg_monetary_by_freq = df.groupby('Frequency')['Monetary'].mean().reset_index()

plt.figure(figsize=(20, 5))
sns.lineplot(data=avg_monetary_by_freq, x='Frequency', y='Monetary', marker='o')
plt.title('Tren Rata-rata Monetary berdasarkan Frequency')
plt.xlabel('Frequency (jumlah transaksi)')
plt.ylabel('Rata-rata Monetary')
plt.show()
```

### **Korelasi Antar Fitur Utama**

---

Menampilkan korelasi antar empat fitur utama saja, bukan seluruh kolom numerik, agar lebih mudah menangkap pola yang relevan.

```python
rfm_features = ['Recency', 'Frequency', 'Monetary', 'Avg_Order_Value']
corr_matrix = df[rfm_features].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap - Fitur RFM')
plt.show()
```
